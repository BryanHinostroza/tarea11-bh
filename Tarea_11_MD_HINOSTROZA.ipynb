a. Calcule las aptitudes y con base a las probabilidades de selección 0.10, 0.60, 0.55 y 0.78, obtenga los nuevos cromosomas. Además, calcule la aptitud promedio en esta fase.


```python
import numpy as np
```

```python
# Datos iniciales
cromosomas_binarios = ['01101', '11100', '01000', '10011']
probabilidades_seleccion = [0.10, 0.60, 0.55, 0.78]
```

```python
# Paso 1: Convertir a decimal
valores_decimales = [int(c, 2) for c in cromosomas_binarios]
```

```python
# Paso 2: Calcular función objetivo y aptitud inversa
def funcion_objetivo(x):
    return x**2 - 2*x + 5
```

```python
# Calcular f(x) y aptitudes (aptitud inversa porque minimizamos)
fx = [funcion_objetivo(x) for x in valores_decimales]
aptitudes = [1 / f for f in fx]  # Inversa para minimizar
```

```python
# Paso 3: Calcular suma total y probabilidad acumulada
suma_aptitudes = sum(aptitudes)
prob_acumulada = np.cumsum([a / suma_aptitudes for a in aptitudes])
```

```python
# Paso 4: Selección de nuevos cromosomas con ruleta
nuevos_cromosomas = []
for p in probabilidades_seleccion:
    for i, prob in enumerate(prob_acumulada):
        if p <= prob:
            nuevos_cromosomas.append(cromosomas_binarios[i])
            break
```

```python
# Paso 5: Aptitud promedio de la nueva generación
nuevos_valores = [int(c, 2) for c in nuevos_cromosomas]
nuevos_fx = [funcion_objetivo(x) for x in nuevos_valores]
aptitud_promedio = sum(nuevos_fx) / len(nuevos_fx)
```

```python
# Mostrar resultados
print("Cromosomas originales:", cromosomas_binarios)
print("Valores decimales:", valores_decimales)
print("f(x):", fx)
print("Aptitudes:", aptitudes)
print("Probabilidad acumulada:", prob_acumulada)
print("Nuevos cromosomas seleccionados:", nuevos_cromosomas)
print("Aptitud promedio de nueva generación:", round(aptitud_promedio, 2))
```

b. Realice un cruce con los dos primeros cromosomas con punto de corte entre la cuarta y quinta posición y un cruce entre los dos últimos cromosomas con punto de corte entre la segunda y tercera posición. También, calcule la aptitud promedio e indique si hubo mejoras.


```python
# Cromosomas seleccionados tras la ruleta
cromosomas = ['01101', '01000', '01000', '01000']
```

```python
# Cruce 1: entre el 1° y 2° cromosoma, corte entre 4° y 5°
p1, p2 = cromosomas[0], cromosomas[1]
hijo1 = p1[:4] + p2[4:]
hijo2 = p2[:4] + p1[4:]
```

```python
# Cruce 2: entre el 3° y 4° cromosoma, corte entre 2° y 3°
p3, p4 = cromosomas[2], cromosomas[3]
hijo3 = p3[:2] + p4[2:]
hijo4 = p4[:2] + p3[2:]
```

```python
# Nueva población después de cruce
nueva_generacion = [hijo1, hijo2, hijo3, hijo4]
```

```python
# Función objetivo
def f(x):
    return x**2 - 2*x + 5

# Calcular valores decimales y f(x)
valores_decimales = [int(c, 2) for c in nueva_generacion]
fx = [f(x) for x in valores_decimales]

# Aptitud promedio
aptitud_promedio_nueva = sum(fx) / len(fx)

# Aptitud promedio anterior (de cromosomas seleccionados)
valores_anteriores = [int(c, 2) for c in cromosomas]
fx_anteriores = [f(x) for x in valores_anteriores]
aptitud_promedio_anterior = sum(fx_anteriores) / len(fx_anteriores)
```

```python
# Resultados
print("Cromosomas después del cruce:", nueva_generacion)
print("Valores decimales:", valores_decimales)
print("f(x):", fx)
print("Aptitud promedio anterior:", round(aptitud_promedio_anterior, 2))
print("Aptitud promedio nueva:", round(aptitud_promedio_nueva, 2))

if aptitud_promedio_nueva < aptitud_promedio_anterior:
    print("Hubo mejora en la aptitud promedio.")
else:
    print("No hubo mejora en la aptitud promedio.")
```

c. Realice una mutación en la primera posición del primer cromosoma y en la tercera posición del último cromosoma. Vuelva a calcular la aptitud promedio e indique si hubo mejoras.

```python
# Cromosomas después del cruce
cromosomas_cruce = ['01100', '01001', '01000', '01000']

# Función para mutar un bit en cierta posición
def mutar(cromosoma, posicion):
    lista = list(cromosoma)
    lista[posicion] = '0' if lista[posicion] == '1' else '1'
    return ''.join(lista)
```

```python
# Mutaciones:
# 1. Primera posición del primer cromosoma (posición 0)
cromosomas_cruce[0] = mutar(cromosomas_cruce[0], 0)

# 2. Tercera posición del último cromosoma (posición 2)
cromosomas_cruce[-1] = mutar(cromosomas_cruce[-1], 2)

# Nueva población después de mutación
nueva_generacion_mutada = cromosomas_cruce
```

```python
# Función objetivo
def f(x):
    return x**2 - 2*x + 5

# Calcular valores decimales y f(x)
valores_decimales = [int(c, 2) for c in nueva_generacion_mutada]
fx = [f(x) for x in valores_decimales]

# Calcular aptitud promedio después de mutación
aptitud_promedio_mutada = sum(fx) / len(fx)

# Calcular aptitud promedio anterior (post-cruce)
valores_post_cruce = [int(c, 2) for c in ['01100', '01001', '01000', '01000']]
fx_post_cruce = [f(x) for x in valores_post_cruce]
aptitud_promedio_post_cruce = sum(fx_post_cruce) / len(fx_post_cruce)
```

```python
# Mostrar resultados
print("Cromosomas después de mutación:", nueva_generacion_mutada)
print("Valores decimales:", valores_decimales)
print("f(x):", fx)
print("Aptitud promedio antes de mutación:", round(aptitud_promedio_post_cruce, 2))
print("Aptitud promedio después de mutación:", round(aptitud_promedio_mutada, 2))

if aptitud_promedio_mutada < aptitud_promedio_post_cruce:
    print("La mutación mejoró la aptitud promedio.")
else:
    print("La mutación no mejoró la aptitud promedio.")
```
